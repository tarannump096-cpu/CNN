```python
import numpy as np
```

```python
def max_pooling(output_matrix,pooling_value):
    pooling_matrix=np.zeros((pooling_value,pooling_value))
    row=0
    for i in range(0,4,2):
        col=0
        for j in range(0,4,2):
            extract_matrix=output_matrix[i:i+2,j:j+2]
            max_result=np.max(extract_matrix)
            pooling_matrix[row,col]=max_result
            col+=1
        row+=1

    return  pooling_matrix
```

```python
output_matrix=np.arange(0,16).reshape(4,4)
```

```python
pooling_value=2
max_pooling(output_matrix,pooling_value)
```

```python
def min_pooling(output_matrix,pooling_value):
    pooling_matrix=np.zeros((pooling_value,pooling_value))
    row=0
    for i in range(0,4,2):
        col=0
        for j in range(0,4,2):
            extract_matrix=output_matrix[i:i+2,j:j+2]
            max_result=np.min(extract_matrix)
            pooling_matrix[row,col]=max_result
            col+=1
        row+=1

    return  pooling_matrix
```

```python
output_matrix=np.arange(0,16).reshape(4,4)
pooling_value=2
min_pooling(output_matrix,pooling_value)
```

```python
def mean_pooling(output_matrix,pooling_value):
    pooling_matrix=np.zeros((pooling_value,pooling_value))
    row=0
    for i in range(0,4,2):
        col=0
        for j in range(0,4,2):
            extract_matrix=output_matrix[i:i+2,j:j+2]
            max_result=np.mean(extract_matrix)
            pooling_matrix[row,col]=max_result
            col+=1
        row+=1

    return  pooling_matrix
```

```python
output_matrix=np.arange(0,16).reshape(4,4)
pooling_value=2
mean_pooling(output_matrix,pooling_value)
```
