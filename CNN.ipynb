```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.read_csv("/content/Admission_Predict_Ver1.1.csv")
```

```python
df1.drop(columns = ["Serial No."], inplace = True)
```

```python
df1.columns
```

```python
x=df1.drop(columns = ["Chance of Admit "])
```

```python
y=df1["Chance of Admit "]
```

```python
df1.shape
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import MinMaxScaler
ss = MinMaxScaler()
xtrain_scaled = ss.fit_transform(xtrain)
xtest_scaled = ss.transform(xtest)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(64, activation = "relu", input_dim = 7))
model.add(Dense(64, activation = "relu"))
model.add(Dense(1, activation = "linear"))
```

```python
model.summary
```

```python
model.compile(loss = "mean_squared_error", optimizer = "Adam")
```

```python
model.fit(xtrain_scaled, ytrain, epochs = 100)
```

```python
from sklearn.metrics import r2_score
ypred = model.predict(xtest_scaled)
r2_score(ytest, ypred)
```

```python
df2 = pd.read_csv("/content/Iris.csv")
```

```python
x1 = df2.drop(columns = ["Species", "Id"])
y1 = df2["Species"]
```

```python
x1train, x1test, y1train, y1test = train_test_split(x1, y1, test_size=0.2, random_state=42)
x1train_scaled = ss.fit_transform(x1train)
x1test_scaled = ss.transform(x1test)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y1train_le = le.fit_transform(y1train)
y1test_le = le.transform(y1test)
```

```python
model1 = Sequential()
model1.add(Dense(64, activation = "relu", input_dim = 4))
model1.add(Dense(64, activation = "relu"))
model1.add(Dense(3, activation = "softmax"))
```

```python
model1.summary
```

```python
model1.compile(loss = "sparse_categorical_crossentropy", optimizer = "Adam")
```

```python
model1.fit(x1train_scaled, y1train_le, epochs = 100)
```

```python
x1train_scaled.shape
```

```python
y1train_le.shape
```

```python
y1pred = model1.predict(x1test_scaled)
y1pred_classes = np.argmax(y1pred, axis=1)

from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y1test_le, y1pred_classes)
print(f"Accuracy of model1 on the test set: {accuracy}")
```
